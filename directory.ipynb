# Lesson 4: Variables
Learn using variables as memory,
[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/bebeday10/penguin-s-guide-to-python/blob/main/lesson%204%20variables.ipynb)

```python
happy = 10
my_friend = "banana"

print(happy * 23)
print(my_friend*3)
print(f"Pretty cool, {my_friend}?")
print(f"You can use f-strings to say variables, {happy=}")
print(f"{my_friend:-^20}: cool")
print(f"{"absolute cinema":_<20}: you can use a thing, < ^ > and a number")
print(f"{"happy friends":#>20}: nice")
```
