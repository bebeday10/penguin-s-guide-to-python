# They have memory now

Welcome to the *Penguin Library*™. We have this robot, who knows a book.

```python
book = "The penguin and the mountains"
```

It will now tell you *your book*. This is a variable, which you can think of as an item the robot has.

```python
print(book) # prints the book
```

Did you want *this book*? We have plenty other.

```python
books: list[str] = ["Magnificent book", "Penguin fights the book", "A hundred penguins", "Penguin How-To"] # a list of books
```

This is also an item, a *list* of books that you would find here in the library.

What do you want here? The *first* book starts at *0*, the second at 1, and so on.  
Choose your book below.

```python
my_book: list = books[0] # take the book
print(my_book) # print it
```

The robot is making an item named `my_book` which is the item you chose in the `books` item.

You want *another book?*

```python
books.append("new book") # add book
print(books)
```

The list *does* look **ugly**. Maybe we can make it more appetizing?

```python
print(", ".join(i for i in books)) # we join each book with a comma (, ).
```

Let's not get too ahead of ourselves. We can use a for loop for lists, `book` will be the book in books. This will cycle each book in the list.  
An f-string lets the robot say a string and a item at the same time.

```python
for book in books:
    print(f"this book is named {book}")
```

The robot can remove an item in the list using `.pop()`. It can even make an item be the popped item from the list.

```python
popped_book: str = books.pop(0)

print(books) # Magnificent book is now gone. Sorry.
print(popped_book) # print the popped book
```

You can see how many items are in the list with `len`.

```python
print(len(books)) # the length of the book list
```
