# Advanced memory

Welcome to the supermarket. Penguins everywhere will come here to buy multiple things.  
The robot has also evolved.
Please check this out!

```python
products: dict[str, int] = {"apple": 2, "banana": 5, "fish": 3, "bag of chips": 4}
```

This is a dictionary. Today, the robot needs to check the stock of the apples.

```python
print(products["apple"])
```

The robot checked the stock by using ["apple"]. It will say the thing attached with apples.  

But what if it didn't exist?

```python
print(products.get("peanuts", 0))
print(products.get("fish", 0))
```

The robot should *always* use the `.get` method instead of the [] method.

```python
my_item: str = "banana"
print(products.get(my_item, 0))
```

`my_item` is *"banana"* which is *5*.  
The robot can make a new item in the dictionary using the [] method.

```python
products["peanuts"] = 5
print(products)
```

The robot can even use a *For Loop*™.

```python
for product, stock in products.items():
    print(f"{product} has {stock}.")

for product, stock in products.items():
    products[product] += 1

print(products)
```

The first *For Loop*™ prints every product with the stock, and the second adds 1 to every product.

The item's value such as `"peanut": 5` (int), can be used with int stuff, like `+=`, `-=`, `*=` etc. Strings can do strings, lists do lists, dictionaries do dictionaries.

```python
customer = {
    "cash": 2.10, # their money
    "wants to buy": { # what they want to buy
        "banana": 2, # bananas
        "peanuts": 4, # peanuts
    },
    "name": "penguin", # their name
}
```
