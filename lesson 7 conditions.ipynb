# Robot can now check if thing has been met

Robots can now decide.  

Penguins are shocked with this new discovery.

```python
banana = 4

if banana > 2:
    print("banana is more than 2")
```

The robot will only speak if the banana is more than 2.

```python
if banana > 8:
    print("banana is more than 8")

else:
    print("banana is not more than 8")
```

The robot will speak `banana is more than 8` if it is more than 8. If it is not, it will say it is not.

```python
if banana > 2:
    # within these 4 spaces, the code will only run if the if is correct.
    banana += 3
    print(banana)

else:
    # within these 4 spaces, the code will only run if the if is wrong.
    print("i am a banana")
```

Within a tab (`    `) (Four spaces), it will run what's inside the if. You may have seen this with a *For Loop*.

```python
penguin = ["house", "penguin", "banana", "the obvious penguin", "prime penguin for all"]

if "banana" in penguin: # banana is in penguin so it will run
    print("banana is in the penguin.")

if "horse" in penguin: # horse is in penguin so it will not run
    print("horse is in the penguin.")

answer: str = "the obvious penguin" 

if answer in penguin: # answer is in penguin so it will run
    print(f"{answer} is in the penguin.")
```

You can do it for lists, too. If the item is in the list, it will run.

`elif` is a short-form of `else if`. It will happen if the `if`/`elif` above *didn't happen*, and this `if` *did happen*.

```python
if "tremendous banana queen" in penguin:
    print("banana")

elif "penguin's impending doom" in penguin:
    print('saved penguins')

elif "prime penguin for all" in penguin:
    print("we saved the penguin.")

    for i in range(4):
        print(f"we saved {i} penguins")

elif "banana" in penguin:
    print("i am correct i won't run")

else:
    print("you did it!")
    
```

Above showcases a few lines of code.
## The `if`s
1. It is incorrect, so it *didn't* run.
2. It is incorrect, so it *didn't* run.
3. It is correct, so it *did* run.
4. It is correct, but one of the ifs above *did* run, so this if *didn't* run.
5. One of the ifs above *did* run, so this *didn't* run.
