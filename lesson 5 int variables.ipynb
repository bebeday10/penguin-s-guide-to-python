# More memory but numbers

Welcome back to math class.  

Penguins need to do algebra.  

The robot can do it, too.

```python
a: int = 2 # item that is 2

print(a)
```

This is an item called `a`, and it is 2.

```python
penguin: int = a + 4

print(penguin)
```

There is now a penguin that is *a but 4 more*. What's *Penguin*™? Turns out, it's *6*.

```python
penguin += 2

print(penguin)
```

The penguin added 2 to itself to be stronger.

```python
print(a > penguin)
```

This shows if a is more than penguin, which is False.

```python
banana: int = penguin
```

A penguin eats a banana, and suddenly the banana is now a penguin. Math class is always like this.

Here is another penguin, they ate the banana and gained it.

```python
another_penguin = 3 # the penguin is 3
another_penguin += banana # they ate the banana
banana = 0 # banana is now 0 😭

print(f"another penguin is {another_penguin}.") # an f-str lets the robot say a phrase and item at the same time.
```

The robot can use a *For Loop*™ to cycle around a number.

The robot is cycling around `another_penguin`. It may be *Dizzy*™.

The range penguin always starts at 0, so the robot adds 1.

```python
print(another_penguin) # first let's see what another_penguin is
print("---") # put a dash so we aren't confused

for i in range(another_penguin):
    print(i + 1) # range starts at 0, so we add 1 at start at 1

print("this is the end") # show the end
```
